## Regular Expressions Übungsaufgabe

```python
import re

# Suche nach Emailadressen mit Regex

text = 'Hallo Team, bitte kontaktiert john.doe@example.com für Fragen zur neuen Software. Falls es technische Probleme gibt, könnt ihr euch direkt an support@tech-help.org wenden. Für geschäftliche Anfragen steht sales@company.net zur Verfügung. Außerdem ist unser Presseteam unter media@news-site.com erreichbar. Falls ihr persönliche Fragen habt, schreibt mir einfach unter alice123@email-service.com oder meldet euch über unser allgemeines Postfach info@web-platform.io. Beste Grüße, Max'

muster = r"." # ergänzen Sie hier die regex um aus dem obigen Text Emailadressen zu extrahieren

print(re.findall(muster,text))
```
